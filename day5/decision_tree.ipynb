```python
import pandas as pd
df = pd.read_csv("titanic.csv")
```

```python
dir(df)
```

```python
input = df[["Pclass","Sex","Age","Fare"]]
target = df["Survived"]
```

```python
input.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
le = LabelEncoder()
input["Sex"] = le.fit_transform(input["Sex"])
```

```python
input["Age"] = input["Age"].fillna(input["Age"].mean())
```

```python
X_train , X_test , y_train , y_test = train_test_split(input,target,test_size=0.2)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
```

```python
score
```

```python
#sample data for checking the model

sample_data = [[2,1,21,100]]
output = model.predict(sample_data)
output
```

```python
import joblib
joblib.dump(model, "titanic_decision_tree_model.pkl")

print("Model saved successfully!")
```
