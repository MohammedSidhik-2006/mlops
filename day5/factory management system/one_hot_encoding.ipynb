```python
import pandas as pd
```

```python
df = pd.read_csv("ai4i2020.csv")

print(df.head())
```

```python
print(df['Type'].unique())
```

```python
encoded_df = pd.get_dummies(df, columns=['Type'])

print(encoded_df.head())
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=['Type'],
    drop_first=True
)

print(encoded_df.head())
```

Dummy Variable Trap happens when one encoded column can be predicted from other columns.

This creates multicollinearity problems in regression models.

To avoid this, one encoded column is dropped using drop_first=True.
