```python
import cv2
import os

```

```python
video_path = r"D:\video_dataset\forest_fire_2.mp4"

output_folder = "new_forest_fire_frames_2"
os.makedirs(output_folder, exist_ok=True)

```

```python
cap = cv2.VideoCapture(video_path)
```

```python
cap.isOpened()
```

```python
frame_count = 0
saved_count = 0
```

```python
while True:
    success, frame = cap.read()

    if not success:
        break

    frame = cv2.resize(frame, (1280, 720))

    # Save every 10th frame
    if frame_count % 10 == 0:

        filename = os.path.join(
            output_folder,
            f"frame_{saved_count}.jpg"
        )

        cv2.imwrite(filename, frame)

        print(f"Saved: {filename}")

        saved_count += 1

    frame_count += 1
```
